# Data Cleaning

## Cleaning decisions based on this file:

### Remove Status and unnamed1 because every value in both columns is missing.
### Remove duplicate records.
### Remove rows where Amount is missing because the sales amount cannot be analyzed for those rows.
### Keep the remaining columns and records; do not invent missing values.

```python
df_clean = df.copy()

# Remove columns that are completely empty
empty_columns = [col for col in df_clean.columns if df_clean[col].isna().all()]
df_clean.drop(columns=empty_columns, inplace=True)
```

```python
# Remove duplicate rows
df_clean.drop_duplicates(inplace=True)
```

```python
# Convert Amount to numeric and remove rows where Amount is missing/invalid
df_clean["Amount"] = pd.to_numeric(df_clean["Amount"], errors="coerce")
df_clean.dropna(subset=["Amount"], inplace=True)
```

```python
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
print("Remaining missing values:")
display(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
print("Remaining duplicates:", df_clean.duplicated().sum())
```
